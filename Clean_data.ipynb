```python
file = open("unclean_smartwatch_health_data.csv")

Data = file.readlines()

Data = [i[:-1:].split(',') for i in Data]
Data = [i[:-1:] for i in Data if ('' not in i
                                  and len(i) == 7
                                  and 'nan' not in i
                                  and 'ERROR' not in i)]
# Sedentary = 0
# Active = 1
# Highly Active = 2
labels = Data[0]
Data = [i[:-1:] + [str(1 if 'Active' in i or 'Actve' in i
                   else 0 if i[5] in ['Seddentary', 'Sedentary']
                    else 2)] for i in Data[1::]]
Data = [i[1::] for i in Data][:2001:]

file.close()
len(Data)
```

```python
file = open('cleaned_dataset.txt', 'w')
#file.write(','.join(labels) + '\n')
for i in Data:
    file.write(','.join(i) + '\n')
file.close()
```
